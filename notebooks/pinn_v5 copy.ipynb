```python
# 🔁 Version V7 - PINN with time-dependent D(t), α(t), and n(t) - Optimized for Speed

import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.optimizers import Adam
import time # For measuring execution time

# 📥 Charger le dataset
df_all = pd.read_csv("Dataset_multi-temps_format_.csv")

# 📊 Calcul automatics des moyennes et écarts-types
input_columns = ['phi', 'T', 'w_c', 'S0', 't_h']
X_all = df_all[input_columns].drop_duplicates().values.astype(np.float32)
X_mean = tf.constant(np.mean(X_all, axis=0), dtype=tf.float32)
X_std = tf.constant(np.std(X_all, axis=0), dtype=tf.float32)
X_std_safe = tf.where(X_std == 0.0, tf.ones_like(X_std), X_std)

# === 2. Normalisation ===
def normalize(x):
    return (x - X_mean) / X_std_safe

# === 3. Time-dependent parameter networks ===
def build_time_param_network(name, output_range):
    """
    Build a small network that takes normalized time as input and outputs a parameter value
    output_range: tuple (min_val, max_val) for the parameter
    """
    model = Sequential([
        Input(shape=(1,), name=f"{name}_time_input"),
        Dense(16, activation='tanh', name=f"{name}_hidden1"),
        Dense(16, activation='tanh', name=f"{name}_hidden2"),
        Dense(1, activation='linear', name=f"{name}_raw_output")
    ], name=f"{name}_network")
    return model, output_range

# Create time-dependent networks for D, α, and n
D_network, D_range = build_time_param_network("D", (1e-9, 1e-7))
alpha_network, alpha_range = build_time_param_network("alpha", (0.005, 0.15))
n_network, n_range = build_time_param_network("n", (1.2, 8.0))

# === 4. Main model that outputs base scaling factors ===
def build_main_model(input_dim=5):
    """
    Main model that takes material properties and outputs base scaling factors
    for the time-dependent parameters
    """
    model = Sequential([
        Input(shape=(input_dim,)),
        Dense(20, activation='tanh'),
        Dense(20, activation='tanh'),
        Dense(3, activation='linear')  # 3 outputs: base scales for D, α, n
    ], name="main_model")
    return model

main_model = build_main_model()

# === 5. Parameter extraction with time dependence ===
@tf.function
def get_time_dependent_params(t_normalized, base_scales, D_net, alpha_net, n_net):
    """
    Get parameter values at normalized time t
    t_normalized: normalized time in [0, 1]
    base_scales: [D_scale, alpha_scale, n_scale] from main model
    """
    # Ensure t_normalized has proper shape
    t_input = tf.reshape(t_normalized, (-1, 1))
    
    # Get raw outputs from time networks
    D_time_factor = D_net(t_input)
    alpha_time_factor = alpha_net(t_input)
    n_time_factor = n_net(t_input)
    
    # Apply sigmoid to time factors to get values in [0, 1]
    D_time_factor = tf.nn.sigmoid(D_time_factor)
    alpha_time_factor = tf.nn.sigmoid(alpha_time_factor)
    n_time_factor = tf.nn.sigmoid(n_time_factor)
    
    # Extract base scales
    D_scale = base_scales[0:1]
    alpha_scale = base_scales[1:2]
    n_scale = base_scales[2:3]
    
    # Compute final parameters with constraints
    # D: exponential scaling around base value (original multiplier, hoping better dt/Nx helps)
    D_base = tf.nn.sigmoid(D_scale) * (D_range[1] - D_range[0]) + D_range[0]
    # Reverted to original 0.5 for more variation, adjust if instability returns
    Dv = D_base * tf.exp((D_time_factor - 0.5) * 0.5)
    Dv = tf.clip_by_value(Dv, D_range[0], D_range[1])
    
    # α: linear scaling around base value (original multiplier)
    alpha_base = tf.nn.sigmoid(alpha_scale) * (alpha_range[1] - alpha_range[0]) + alpha_range[0]
    # Reverted to original 0.6 for more variation, adjust if instability returns
    alpha0 = alpha_base * (0.7 + 0.6 * alpha_time_factor)
    alpha0 = tf.clip_by_value(alpha0, alpha_range[0], alpha_range[1])
    
    # n: linear scaling around base value (original multiplier)
    n_base = tf.nn.sigmoid(n_scale) * (n_range[1] - n_range[0]) + n_range[0]
    # Reverted to original 0.4 for more variation, adjust if instability returns
    nShape = n_base * (0.8 + 0.4 * n_time_factor)
    nShape = tf.clip_by_value(nShape, n_range[0], n_range[1])
    
    return tf.squeeze(Dv), tf.squeeze(alpha0), tf.squeeze(nShape)

# === 6. Modified simulation with time-dependent parameters ===
@tf.function
def simulate_model_tf_time_dependent(base_scales, D_net, alpha_net, n_net, 
                                     S0, phi, rho_eau, dx, dt, Tmax, Nx, mode="imbibition"):
    """
    Simulation with time-dependent parameters
    """
    # Convert scalars to tensors
    dx = tf.convert_to_tensor(dx, dtype=tf.float32)
    dt = tf.convert_to_tensor(dt, dtype=tf.float32)
    Tmax = tf.convert_to_tensor(Tmax, dtype=tf.float32)
    Nx = tf.convert_to_tensor(Nx, tf.int32) 

    Nt = tf.cast(tf.math.floordiv(Tmax, dt), tf.int32)
    x_vals = tf.linspace(0.0, dx * tf.cast(Nx - 1, tf.float32), Nx)

    # Physical parameters
    Sc = 0.8
    S_cap = 0.75
    beta = 2.5
    Pc0 = 5e6
    K_in = 1.7e-20
    eta_l = 1e-3

    # Initialize saturation
    if mode == "imbibition":
        S = tf.ones((Nx,), dtype=tf.float32) * 0.01
        S = tf.tensor_scatter_nd_update(S, [[0]], [S0])
    elif mode == "drainage":
        S = tf.ones((Nx,), dtype=tf.float32) * S0
    else:
        raise ValueError("Mode must be 'imbibition' or 'drainage'")

    M = tf.ones((Nx,), dtype=tf.float32) * dx
    M = tf.tensor_scatter_nd_update(M, [[0], [Nx - 1]], [dx / 2, dx / 2])

    # Define diffusion functions outside the loop
    def Dvapeur(S_val, alpha0, nShape, Dv):
        S_val = tf.clip_by_value(S_val, 1e-4, 1.0)
        fs = alpha0 + (1.0 - alpha0) / (1.0 + tf.pow((1.0 - S_val) / (1.0 - Sc), nShape))
        return tf.clip_by_value(fs, 1e-12, 10.0) * Dv

    def Dcapi(S_val):
        S_val = tf.clip_by_value(S_val, 1e-4, 1.0)
        dpc = -beta * Pc0 * tf.pow(S_val, -beta - 1)
        return tf.abs(dpc) * K_in / eta_l

    def D_S(S_val, alpha0, nShape, Dv):
        weight = tf.math.sigmoid(50.0 * (S_val - S_cap))
        return weight * Dcapi(S_val) + (1.0 - weight) * Dvapeur(S_val, alpha0, nShape, Dv)

    # Time loop with time-dependent parameters
    current_time = tf.constant(0.0)
    
    for i in tf.range(Nt):
        # Normalize time to [0, 1]
        t_normalized = current_time / Tmax
        
        # Get time-dependent parameters
        Dv, alpha0, nShape = get_time_dependent_params(t_normalized, base_scales, 
                                                       D_net, alpha_net, n_net)
        
        S_old = S
        # Reverted to 5 Picard iterations for speed. Adjust to 10-20 if instability returns.
        for _ in tf.range(5):  # Picard iterations (Reverted from 10 to 5)
            D_vals = D_S(S, alpha0, nShape, Dv)

            D_left = (D_vals[:-1] + D_vals[1:]) / 2.0
            fluxes = D_left * (S[1:] - S[:-1]) / dx

            dS_dt = tf.zeros_like(S)

            # Update interior points
            interior_flux = (fluxes[1:] - fluxes[:-1]) / dx
            dS_dt = tf.tensor_scatter_nd_add(
                dS_dt,
                tf.range(1, Nx - 1)[:, tf.newaxis],
                interior_flux
            )

            # Right boundary: Neumann ≈ 0
            last_flux = fluxes[-1] / dx
            dS_dt = tf.tensor_scatter_nd_add(dS_dt, [[Nx - 1]], [-last_flux])

            # Update S
            S = S_old + dt * dS_dt
            S = tf.tensor_scatter_nd_update(S, [[0]], [S0])  # Dirichlet
        
        # Update time
        current_time = current_time + dt

    w_final = phi * rho_eau * S
    return x_vals, w_final

# === 7. Combined optimizer for all networks ===
all_models = [main_model, D_network, alpha_network, n_network]
# Set learning rate to a balanced value. 1e-3 is often standard, but can go lower.
optimizer = Adam(learning_rate=5e-4) # Slightly increased from 1e-4 to 5e-4, but still conservative

# === 8. Modified training function with Batching ===
@tf.function
def tf_linear_interp(x_known, y_known, x_interp):
    x_interp_m = x_interp * 1e-3
    indices = tf.searchsorted(x_known, x_interp_m, side='right') - 1
    indices = tf.clip_by_value(indices, 0, tf.shape(x_known)[0] - 2)
    x0 = tf.gather(x_known, indices)
    x1 = tf.gather(x_known, indices + 1)
    y0 = tf.gather(y_known, indices)
    y1 = tf.gather(y_known, indices + 1)
    return y0 + (y1 - y0) * (x_interp_m - x0) / (x1 - x0 + 1e-12)

def train_epoch_time_dependent(all_models, df_all, optimizer, normalize, 
                               simulate_model_tf_time_dependent, tf_linear_interp, 
                               Nx_val, dx_val, dt_val, batch_size=None): # Added batch_size
    
    main_model, D_net, alpha_net, n_net = all_models
    
    unique_inputs_df = df_all[['phi', 'T', 'w_c', 'S0', 't_h']].drop_duplicates()
    unique_inputs_values = unique_inputs_df.values.astype(np.float32)

    # Shuffle and create batches
    if batch_size is None:
        batch_size = len(unique_inputs_values) # Process all at once if batch_size is None
    
    # Convert to tf.data.Dataset for easy batching and shuffling
    dataset = tf.data.Dataset.from_tensor_slices(unique_inputs_values)
    dataset = dataset.shuffle(buffer_size=len(unique_inputs_values)).batch(batch_size)

    total_loss = 0.0
    valid_count = 0
    param_epoch_vals = {'Dv_mean': [], 'alpha_mean': [], 'n_mean': []}

    with tf.GradientTape() as tape:
        for batch_idx, batch_inputs_np in enumerate(dataset):
            batch_inputs = tf.cast(batch_inputs_np, tf.float32) # Ensure float32

            # Iterate through each item in the batch
            for j in range(batch_inputs.shape[0]):
                phi, T, wc, S0, t_h = batch_inputs[j]
                input_tensor = tf.reshape([phi, T, wc, S0, t_h], (1, -1)) # Reshape for normalize

                norm_input = normalize(input_tensor)
                if tf.reduce_any(tf.math.is_nan(norm_input)):
                    # print(f"❌ NaN in normalized input for batch {batch_idx}, item {j}")
                    continue
                
                # Get base scales from main model
                base_scales = main_model(norm_input)[0]
                
                if tf.reduce_any(tf.math.is_nan(base_scales)):
                    # print(f"❌ NaN in model output for batch {batch_idx}, item {j}")
                    continue

                # Get experimental data
                # Fetching experimental data still needs to be done per-profile
                phi_f, T_f, wc_f, S0_f, t_h_f = map(float, [phi, T, wc, S0, t_h])
                group = df_all[
                    (df_all['phi'].round(4) == round(phi_f, 4)) &
                    (df_all['T'] == T_f) &
                    (df_all['w_c'].round(4) == round(wc_f, 4)) &
                    (df_all['S0'].round(4) == round(S0_f, 4)) &
                    (df_all['t_h'] == t_h_f)
                ]

                x_exp = tf.constant(group['x_mm'].values.astype(np.float32))
                w_exp = tf.constant(group['w_exp'].values.astype(np.float32))

                if len(x_exp) < 3:
                    # print("❌ Too few points for profile:", t_h)
                    continue

                w_exp_mean = tf.reduce_mean(w_exp)
                w_exp_std = tf.math.reduce_std(w_exp)
                if tf.math.is_nan(w_exp_std) or w_exp_std < 1e-8:
                    # print(f"❌ Rejected: std(w_exp) = {w_exp_std.numpy()}")
                    continue

                try:
                    # Run simulation with time-dependent parameters
                    x_model, w_model = simulate_model_tf_time_dependent(
                        base_scales, D_net, alpha_net, n_net,
                        S0, phi, 1000.0, dx_val, dt_val, t_h * 3600.0, Nx_val
                    )
                    
                    if tf.reduce_any(tf.math.is_nan(w_model)):
                        # print("❌ NaN in simulation output (w_model)")
                        continue
                        
                except Exception as e:
                    # print(f"❌ Simulation error: {e}")
                    continue
                
                # Interpolate and compute loss
                x_exp_clip = tf.clip_by_value(x_exp * 1e-3, x_model[0], x_model[-1]) * 1e3
                w_interp = tf_linear_interp(x_model, w_model, x_exp_clip)

                w_exp_norm = (w_exp - w_exp_mean) / (w_exp_std + 1e-12)
                w_interp_norm = (w_interp - w_exp_mean) / (w_exp_std + 1e-12)

                if tf.reduce_any(tf.math.is_nan(w_interp_norm)):
                    # print("❌ Rejected: NaN in w_interp_norm (after interpolation)")
                    continue

                loss = tf.reduce_mean(tf.square(w_interp_norm - w_exp_norm))
                if tf.math.is_nan(loss):
                    # print("❌ Rejected: NaN in loss calculation")
                    continue

                # print(f"✅ Profile {batch_idx*batch_size + j+1}/{len(unique_inputs_values)} accepted. Loss = {loss.numpy():.4e}")

                total_loss += loss
                valid_count += 1
                
                # Sample parameters at different times for monitoring
                for t_sample in [0.0, 0.5, 1.0]:
                    t_norm = tf.constant(t_sample)
                    Dv_t, alpha_t, n_t = get_time_dependent_params(t_norm, base_scales, 
                                                                   D_net, alpha_net, n_net)
                    param_epoch_vals['Dv_mean'].append(Dv_t.numpy())
                    param_epoch_vals['alpha_mean'].append(alpha_t.numpy())
                    param_epoch_vals['n_mean'].append(n_t.numpy())

        if valid_count == 0:
            print("❌ No valid profiles processed in this epoch. Adjust batch size or data filtering.")
            return None, None

        mean_loss = total_loss / valid_count

    # Compute gradients for all models
    all_variables = []
    for model in all_models:
        all_variables.extend(model.trainable_variables)
    
    grads = tape.gradient(mean_loss, all_variables)
    grads = [tf.clip_by_value(g, -0.5, 0.5) if g is not None else None for g in grads]
    
    # Check for NaN in gradients
    grads_clean = []
    for g in grads:
        if g is not None and tf.reduce_any(tf.math.is_nan(g)):
            print("⚠️ NaN in gradients, replaced with zeros")
            grads_clean.append(tf.zeros_like(g))
        else:
            grads_clean.append(g)
    
    optimizer.apply_gradients(zip(grads_clean, all_variables))

    mean_params = {k: np.mean(v) if v else 0.0 for k, v in param_epoch_vals.items()}
    return mean_loss.numpy(), mean_params

# === 9. Visualization function for time-dependent parameters ===
def visualize_time_dependent_params(base_scales, D_net, alpha_net, n_net, title_suffix=""):
    """
    Visualize how parameters change over time
    """
    time_points = np.linspace(0, 1, 100)
    D_values = []
    alpha_values = []
    n_values = []
    
    for t in time_points:
        t_norm = tf.constant(t, dtype=tf.float32)
        Dv, alpha0, nShape = get_time_dependent_params(t_norm, base_scales, 
                                                       D_net, alpha_net, n_net)
        D_values.append(Dv.numpy())
        alpha_values.append(alpha0.numpy())
        n_values.append(nShape.numpy())
    
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    
    axes[0].plot(time_points, D_values, 'b-', linewidth=2)
    axes[0].set_xlabel('Normalized Time')
    axes[0].set_ylabel('D (m²/s)')
    axes[0].set_title(f'D(t) {title_suffix}')
    axes[0].grid(True)
    axes[0].ticklabel_format(style='sci', axis='y', scilimits=(0,0))
    
    axes[1].plot(time_points, alpha_values, 'g-', linewidth=2)
    axes[1].set_xlabel('Normalized Time')
    axes[1].set_ylabel('α')
    axes[1].set_title(f'α(t) {title_suffix}')
    axes[1].grid(True)
    
    axes[2].plot(time_points, n_values, 'r-', linewidth=2)
    axes[2].set_xlabel('Normalized Time')
    axes[2].set_ylabel('n')
    axes[2].set_title(f'n(t) {title_suffix}')
    axes[2].grid(True)
    
    plt.tight_layout()
    plt.show()

# === 10. Training loop ===
loss_history = []
param_history = {'Dv_mean': [], 'alpha_mean': [], 'n_mean': []}

print("🚀 Starting training with time-dependent parameters...")

# --- Tuned parameters for speed and stability ---
# Start with a dt that is a balance. 10.0-30.0 is often a good range.
# Nx can be lower if dt is also lower.
OPTIMAL_NX = 60 # Reduced from 100, increased from 40
OPTIMAL_DT = 15.0 # Increased from 1.0, reduced from 60.0
BATCH_SIZE = 1 # Keep at 1 for now if issues with larger batches, else try 4, 8, etc.

for epoch in range(100):
    epoch_start_time = time.time() # Start timing for the epoch

    mean_loss, mean_params = train_epoch_time_dependent(
        all_models, df_all, optimizer, normalize,
        simulate_model_tf_time_dependent, tf_linear_interp,
        Nx_val=OPTIMAL_NX, dx_val=1e-3, dt_val=OPTIMAL_DT,
        batch_size=BATCH_SIZE # Use batching
    )

    epoch_end_time = time.time() # End timing for the epoch

    if mean_loss is None:
        print(f"Epoch {epoch}: ❌ No valid profiles processed → training stopped.")
        break

    loss_history.append(mean_loss)
    for key, value in mean_params.items():
        param_history[key].append(value)

    if epoch % 1 == 0: # Print every epoch to see progress quickly
        print(f"\n📉 Epoch {epoch} took {epoch_end_time - epoch_start_time:.2f}s, Loss: {mean_loss:.4e}")
        print(f"   Mean values: D={mean_params['Dv_mean']:.2e}, α={mean_params['alpha_mean']:.4f}, n={mean_params['n_mean']:.2f}")
        
        # Visualize time-dependent parameters for a test case less frequently (e.g., every 10 epochs)
        if epoch % 10 == 0:
            test_input = tf.constant([[0.119, 20.0, 0.45, 0.6, 72.0]], dtype=tf.float32)
            test_base_scales = main_model(normalize(test_input))[0]
            visualize_time_dependent_params(test_base_scales, D_network, alpha_network, n_network, 
                                            title_suffix=f"(Epoch {epoch})")
            
            # Save models less frequently
            main_model.save_weights(f"main_model_timedep_epoch{epoch}.weights.h5")
            D_network.save_weights(f"D_network_epoch{epoch}.weights.h5")
            alpha_network.save_weights(f"alpha_network_epoch{epoch}.weights.h5")
            n_network.save_weights(f"n_network_epoch{epoch}.weights.h5")

# === 11. Final convergence plots ===
if loss_history:
    fig, axes = plt.subplots(2, 2, figsize=(12, 8))

    axes[0,0].plot(loss_history, label="Loss", linewidth=2, color='red')
    axes[0,0].set_xlabel("Epoch")
    axes[0,0].set_ylabel("Loss")
    axes[0,0].set_title("Loss Convergence")
    axes[0,0].grid(True)
    axes[0,0].legend()
    axes[0,0].set_yscale('log')

    if param_history['Dv_mean']:
        axes[0,1].plot(param_history['Dv_mean'], label="D mean", linewidth=2, color='blue')
        axes[0,1].set_xlabel("Epoch")
        axes[0,1].set_ylabel("D (m²/s)")
        axes[0,1].set_title("Mean D Evolution")
        axes[0,1].grid(True)
        axes[0,1].legend()

    if param_history['alpha_mean']:
        axes[1,0].plot(param_history['alpha_mean'], label="α mean", linewidth=2, color='green')
        axes[1,0].set_xlabel("Epoch")
        axes[1,0].set_ylabel("α")
        axes[1,0].set_title("Mean α Evolution")
        axes[1,0].grid(True)
        axes[1,0].legend()

    if param_history['n_mean']:
        axes[1,1].plot(param_history['n_mean'], label="n mean", linewidth=2, color='orange')
        axes[1,1].set_xlabel("Epoch")
        axes[1,1].set_ylabel("n")
        axes[1,1].set_title("Mean n Evolution")
        axes[1,1].grid(True)
    axes[1,1].legend()

    plt.tight_layout()
    plt.show()

print("🎉 Training completed! The model now uses time-dependent D(t), α(t), and n(t).")

# === 12. Test the trained model ===
print("\n🧪 Testing time-dependent model...")
test_input = tf.constant([[0.119, 20.0, 0.45, 0.6, 72.0]], dtype=tf.float32)
test_base_scales = main_model(normalize(test_input))[0]

# Show parameter evolution
visualize_time_dependent_params(test_base_scales, D_network, alpha_network, n_network, 
                                title_suffix="(Final Model)")

# Run simulation and plot result
x_model, w_model = simulate_model_tf_time_dependent(
    test_base_scales, D_network, alpha_network, n_network,
    0.6, 0.119, 1000.0, dx=1e-3, dt=30.0, Tmax=72*3600, Nx=100
)

plt.figure(figsize=(8, 5))
plt.plot(x_model.numpy()*1000, w_model.numpy(), label="Simulation with time-dependent params", linewidth=2)
plt.title("Test Case - t=72h with Time-Dependent Parameters")
plt.xlabel("x [mm]")
plt.ylabel("w [kg/m³]")
plt.grid()
plt.legend()
plt.show()
```

```python
import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.optimizers import Adam
import time

# --- Global Calibrator Variables ---
# These will be learnable during training. Initialize as None.
_D_calibrator_scale = None
_D_calibrator_offset = None
_alpha_calibrator_scale = None
_alpha_calibrator_offset = None
_n_calibrator_scale = None
_n_calibrator_offset = None

# --- Hyperparameters for the PINN (tuned for speed) ---
OPTIMAL_NX = 50  # Reduced from 60
OPTIMAL_DT = 30.0 # Increased from 15.0 (can go higher, e.g., 45.0, 60.0, if stable)
PICARD_ITERATIONS = 3 # Reduced from 5 (can go to 2 if needed)
BATCH_SIZE = 8 # Increased from 1 (crucial for GPU usage, start low, increase gradually)
EPOCHS = 200 # Can adjust as needed, but faster epochs mean more iterations in same time

# 📥 Load the dataset
df_all = pd.read_csv("Dataset_multi-temps_format_.csv")

# 📊 Calculate mean and std for normalization
input_columns = ['phi', 'T', 'w_c', 'S0', 't_h']
X_all_np = df_all[input_columns].drop_duplicates().values.astype(np.float32)
X_mean = tf.constant(np.mean(X_all_np, axis=0), dtype=tf.float32)
X_std = tf.constant(np.std(X_all_np, axis=0), dtype=tf.float32)
X_std_safe = tf.where(X_std == 0.0, tf.ones_like(X_std), X_std)

# === 2. Normalization ===
def normalize_input_features(x):
    return (x - X_mean) / X_std_safe

# === 3. Time-dependent parameter networks ===
def build_time_param_network(name): # Removed output_range, handled by calibrator
    """
    Build a small network that takes normalized time as input and outputs a raw parameter value
    """
    model = Sequential([
        Input(shape=(1,), name=f"{name}_time_input"),
        Dense(10, activation='tanh', name=f"{name}_hidden1"), # Reduced neurons
        Dense(10, activation='tanh', name=f"{name}_hidden2"), # Reduced neurons
        Dense(1, activation='linear', name=f"{name}_raw_output")
    ], name=f"{name}_network")
    return model

# Define target ranges for parameters (used by calibrator)
D_range = (1e-10, 1e-7) # Adjusted D_range for more realistic values, was 1e-9 to 1e-7, made slightly wider lower bound
alpha_range = (0.005, 0.15)
n_range = (1.2, 8.0)

# Create time-dependent networks for D, α, and n
D_network = build_time_param_network("D")
alpha_network = build_time_param_network("alpha")
n_network = build_time_param_network("n")

# === 4. Main model that outputs base scaling factors ===
def build_main_model(input_dim=5):
    """
    Main model that takes material properties and outputs base scaling factors
    for the time-dependent parameters
    """
    model = Sequential([
        Input(shape=(input_dim,)),
        Dense(15, activation='tanh'), # Reduced neurons
        Dense(15, activation='tanh'), # Reduced neurons
        Dense(3, activation='linear')  # 3 outputs: base scales for D, α, n
    ], name="main_model")
    return model

main_model = build_main_model()

# === 5. Parameter extraction with time dependence (WITH CALIBRATORS) ===
@tf.function
def get_time_dependent_params(t_normalized, base_scales):
    """
    Get parameter values at normalized time t, with learnable calibrators
    """
    global _D_calibrator_scale, _D_calibrator_offset
    global _alpha_calibrator_scale, _alpha_calibrator_offset
    global _n_calibrator_scale, _n_calibrator_offset

    # Initialize calibrator variables on first call (within tf.function context)
    if _D_calibrator_scale is None: # This check ensures variables are created only once
        _D_calibrator_scale = tf.Variable(0.1, dtype=tf.float32, name='D_cal_scale')
        _D_calibrator_offset = tf.Variable(0.0, dtype=tf.float32, name='D_cal_offset')
        _alpha_calibrator_scale = tf.Variable(0.1, dtype=tf.float32, name='alpha_cal_scale')
        _alpha_calibrator_offset = tf.Variable(0.0, dtype=tf.float32, name='alpha_cal_offset')
        _n_calibrator_scale = tf.Variable(0.1, dtype=tf.float32, name='n_cal_scale')
        _n_calibrator_offset = tf.Variable(0.0, dtype=tf.float32, name='n_cal_offset')

    t_input = tf.reshape(t_normalized, (-1, 1))
    
    # Get raw outputs from time networks
    D_time_factor_raw = D_network(t_input)
    alpha_time_factor_raw = alpha_network(t_input)
    n_time_factor_raw = n_network(t_input)
    
    # Apply sigmoid to raw time factors to get a value typically in [0, 1] for relative trend
    D_time_factor = tf.nn.sigmoid(D_time_factor_raw)
    alpha_time_factor = tf.nn.sigmoid(alpha_time_factor_raw)
    n_time_factor = tf.nn.sigmoid(n_time_factor_raw)
    
    # Extract base scales from main model (also scaled by sigmoid to [0,1])
    # Ensure base_scales is correctly indexed and potentially squeezed for scalar operation
    D_scale = tf.nn.sigmoid(base_scales[0]) # Now just a scalar, not a slice
    alpha_scale = tf.nn.sigmoid(base_scales[1])
    n_scale = tf.nn.sigmoid(base_scales[2])
    
    # Apply Calibrators and map to defined ranges
    D_combined_input = D_scale + (D_time_factor - 0.5) * _D_calibrator_scale + _D_calibrator_offset
    Dv = D_range[0] + (D_range[1] - D_range[0]) * tf.nn.sigmoid(D_combined_input)
    # Clipping can act as a hard constraint after sigmoid mapping if needed, but sigmoid should handle range
    Dv = tf.clip_by_value(Dv, D_range[0], D_range[1])
    
    alpha_combined_input = alpha_scale + (alpha_time_factor - 0.5) * _alpha_calibrator_scale + _alpha_calibrator_offset
    alpha0 = alpha_range[0] + (alpha_range[1] - alpha_range[0]) * tf.nn.sigmoid(alpha_combined_input)
    alpha0 = tf.clip_by_value(alpha0, alpha_range[0], alpha_range[1])
    
    n_combined_input = n_scale + (n_time_factor - 0.5) * _n_calibrator_scale + _n_calibrator_offset
    nShape = n_range[0] + (n_range[1] - n_range[0]) * tf.nn.sigmoid(n_combined_input)
    nShape = tf.clip_by_value(nShape, n_range[0], n_range[1])
    
    # Squeeze to remove extra dimensions if present (e.g., from batching within tf.function)
    return tf.squeeze(Dv), tf.squeeze(alpha0), tf.squeeze(nShape)


# === 6. Main Simulation Function (Modified Picard Iterations) ===
@tf.function
def simulate_model_tf_time_dependent(input_params, D_net, alpha_net, n_net, 
                                     dx, dt, Nx, picard_iterations):
    # Unpack input_params (phi, T, wc, S0, t_h, etc.)
    phi = input_params[0]
    S0 = input_params[3]
    Tmax = input_params[4] * 3600.0 # Convert t_h to seconds
    
    rho_eau = 1000.0 # Constant

    dx = tf.cast(dx, dtype=tf.float32)
    dt = tf.cast(dt, dtype=tf.float32)
    Tmax = tf.cast(Tmax, dtype=tf.float32)
    Nx = tf.cast(Nx, tf.int32) 

    Nt = tf.cast(tf.math.floordiv(Tmax, dt), tf.int32)
    x_vals = tf.linspace(0.0, dx * tf.cast(Nx - 1, tf.float32), Nx)

    Sc = 0.8
    S_cap = 0.75
    beta = 2.5
    Pc0 = 5e6
    K_in = 1.7e-20
    eta_l = 1e-3

    S = tf.ones((Nx,), dtype=tf.float32) * 0.01 # Initial uniform low saturation
    S = tf.tensor_scatter_nd_update(S, [[0]], [S0]) # Apply surface saturation boundary condition

    def Dvapeur_tf(S_val_in, alpha0_in, nShape_in, Dv_in):
        S_val_in = tf.clip_by_value(S_val_in, 1e-4, 1.0)
        fs = alpha0_in + (1.0 - alpha0_in) / (1.0 + tf.pow((1.0 - S_val_in) / (1.0 - Sc), nShape_in))
        return tf.clip_by_value(fs, 1e-12, 10.0) * Dv_in

    def Dcapi_tf(S_val_in):
        S_val_in = tf.clip_by_value(S_val_in, 1e-4, 1.0)
        dpc = -beta * Pc0 * tf.pow(S_val_in, -beta - 1)
        return tf.abs(dpc) * K_in / eta_l

    def D_S_tf(S_val_in, alpha0_in, nShape_in, Dv_in):
        weight = tf.math.sigmoid(50.0 * (S_val_in - S_cap))
        return weight * Dcapi_tf(S_val_in) + (1.0 - weight) * Dvapeur_tf(S_val_in, alpha0_in, nShape_in, Dv_in)

    current_time = tf.constant(0.0)
    
    # Get base scales from main_model once per simulation
    input_features_normalized = normalize_input_features(tf.reshape(input_params, (1, -1)))
    base_scales = main_model(input_features_normalized)[0]

    for i in tf.range(Nt):
        t_normalized = current_time / Tmax
        Dv, alpha0, nShape = get_time_dependent_params(t_normalized, base_scales)
        
        S_old = S # Store S from previous time step for Picard iterations
        
        for _ in tf.range(picard_iterations): # Reduced Picard iterations
            D_vals = D_S_tf(S, alpha0, nShape, Dv)

            # Calculate fluxes using central difference for D and forward/backward for S
            # This is a common pattern for FDM diffusion
            D_left = (D_vals[:-1] + D_vals[1:]) / 2.0
            fluxes = D_left * (S[1:] - S[:-1]) / dx # Fick's first law

            dS_dt = tf.zeros_like(S)

            # Apply divergence operator (∂/∂x (D ∂S/∂x))
            # Interior points: (flux_right - flux_left) / dx
            interior_flux_diff = (fluxes[1:] - fluxes[:-1]) / dx
            dS_dt = tf.tensor_scatter_nd_add(
                dS_dt,
                tf.range(1, Nx - 1)[:, tf.newaxis],
                interior_flux_diff
            )

            # Boundary condition at x=L (right boundary, Neumann zero flux)
            # The last flux is -fluxes[-1] / dx
            dS_dt = tf.tensor_scatter_nd_add(dS_dt, [[Nx - 1]], [-fluxes[-1] / dx])

            # Update S
            S = S_old + dt * dS_dt
            
            # Boundary condition at x=0 (left boundary, Dirichlet S0)
            S = tf.tensor_scatter_nd_update(S, [[0]], [S0])
            
        current_time = current_time + dt

    w_final = phi * rho_eau * S
    return x_vals, w_final

@tf.function
def tf_linear_interp(x_known, y_known, x_interp):
    # Ensure x_interp is within bounds of x_known
    x_interp = tf.clip_by_value(x_interp, x_known[0], x_known[-1])

    indices = tf.searchsorted(x_known, x_interp, side='right') - 1
    indices = tf.clip_by_value(indices, 0, tf.shape(x_known)[0] - 2)
    x0 = tf.gather(x_known, indices)
    x1 = tf.gather(x_known, indices + 1)
    y0 = tf.gather(y_known, indices)
    y1 = tf.gather(y_known, indices + 1)
    
    # Avoid division by zero if x0 == x1
    denom = (x1 - x0)
    interp_factor = (x_interp - x0) / (denom + tf.cast(tf.equal(denom, 0.0), tf.float32) * 1e-12)
    return y0 + (y1 - y0) * interp_factor

# --- NEW: Prepare Dataset for Training Loop ---
print("Preparing training dataset...")
processed_profiles = []
unique_inputs_df = df_all[input_columns].drop_duplicates()

for idx, row in unique_inputs_df.iterrows():
    phi_f, T_f, wc_f, S0_f, t_h_f = row['phi'], row['T'], row['w_c'], row['S0'], row['t_h']
    
    group = df_all[
        (df_all['phi'].round(4) == round(phi_f, 4)) &
        (df_all['T'] == T_f) &
        (df_all['w_c'].round(4) == round(wc_f, 4)) &
        (df_all['S0'].round(4) == round(S0_f, 4)) &
        (df_all['t_h'] == t_h_f)
    ]

    x_exp_profile_mm = tf.constant(group['x_mm'].values.astype(np.float32))
    w_exp_profile = tf.constant(group['w_exp'].values.astype(np.float32))

    # Basic filtering of problematic data points
    if len(x_exp_profile_mm) < 3: # Need at least 3 points for meaningful interpolation/std dev
        continue
    
    w_exp_std_val = tf.math.reduce_std(w_exp_profile)
    if tf.math.is_nan(w_exp_std_val) or w_exp_std_val < 1e-6: # Check for very small std or NaN
        continue
    
    # Store input features as a single tensor for direct passing to `simulate_model_tf_time_dependent`
    input_features_tensor = tf.constant([phi_f, T_f, wc_f, S0_f, t_h_f], dtype=tf.float32)
    
    processed_profiles.append((input_features_tensor, x_exp_profile_mm, w_exp_profile))

if not processed_profiles:
    raise ValueError("No valid profiles found after pre-processing. Check your data or filtering criteria.")

# Create the tf.data.Dataset
# Using from_generator because x_exp_profile_mm and w_exp_profile might have varying lengths
def profile_generator():
    for item in processed_profiles:
        yield item

output_signature = (
    tf.TensorSpec(shape=(5,), dtype=tf.float32), # input_features_tensor
    tf.TensorSpec(shape=(None,), dtype=tf.float32), # x_exp_profile_mm (None for variable length)
    tf.TensorSpec(shape=(None,), dtype=tf.float32)  # w_exp_profile (None for variable length)
)

full_dataset = tf.data.Dataset.from_generator(
    profile_generator,
    output_signature=output_signature
)

# Shuffle, batch, cache, and prefetch for optimal performance
# padded_batch is necessary if experimental profiles have different numbers of points
full_dataset = full_dataset.shuffle(buffer_size=len(processed_profiles)).padded_batch(
    BATCH_SIZE,
    padded_shapes=((5,), (None,), (None,)), # Pad x_exp, w_exp to max length in batch
    padding_values=(tf.constant(0., dtype=tf.float32), tf.constant(0., dtype=tf.float32), tf.constant(0., dtype=tf.float32))
).cache().prefetch(tf.data.AUTOTUNE)

print(f"Dataset prepared with {len(processed_profiles)} unique profiles.")

# === 7. Combined optimizer for all networks (NOW INCLUDING CALIBRATOR VARIABLES) ===
all_models = [main_model, D_network, alpha_network, n_network]

# Ensure calibrator variables are created before optimizer is initialized
# Call get_time_dependent_params once with dummy inputs to force variable creation
_ = get_time_dependent_params(tf.constant(0.5), tf.constant([0.5, 0.5, 0.5]))

# Collect all trainable variables: from models and from calibrators
all_trainable_variables = []
for model in all_models:
    all_trainable_variables.extend(model.trainable_variables)
all_trainable_variables.extend([_D_calibrator_scale, _D_calibrator_offset,
                                _alpha_calibrator_scale, _alpha_calibrator_offset,
                                _n_calibrator_scale, _n_calibrator_offset])

optimizer = Adam(learning_rate=5e-4) # Consider reducing if loss oscillates or for fine-tuning

# === 8. Modified training function with Batching ===
@tf.function
def train_step(batch_inputs, batch_x_exp_mm, batch_w_exp):
    total_batch_loss = tf.constant(0.0, dtype=tf.float32)
    valid_profile_count = tf.constant(0, dtype=tf.int32)
    
    # Use tf.GradientTape for the entire batch
    with tf.GradientTape() as tape:
        # Loop through each profile in the batch
        for j in tf.range(tf.shape(batch_inputs)[0]):
            input_features = batch_inputs[j]
            x_exp_profile_mm = batch_x_exp_mm[j]
            w_exp_profile = batch_w_exp[j]

            # Skip padded entries if using padded_batch
            if tf.reduce_all(tf.equal(input_features, 0.0)) and tf.reduce_all(tf.equal(x_exp_profile_mm, 0.0)):
                continue

            # Calculate w_exp_mean and std for normalization
            w_exp_mean = tf.reduce_mean(w_exp_profile)
            w_exp_std = tf.math.reduce_std(w_exp_profile)
            
            # Filter out problematic profiles (very small std or NaN) within the tf.function
            if w_exp_std < 1e-6 or tf.math.is_nan(w_exp_std):
                continue
            
            try:
                x_model_m, w_model = simulate_model_tf_time_dependent(
                    input_features, D_network, alpha_network, n_network, # Pass networks directly
                    dx=1e-3, dt=OPTIMAL_DT, Nx=OPTIMAL_NX, picard_iterations=PICARD_ITERATIONS
                )
                
                if tf.reduce_any(tf.math.is_nan(w_model)):
                    # tf.print("⚠️ NaN in w_model, skipping profile.")
                    continue
                    
            except Exception as e:
                # tf.print("❌ Simulation error during train_step:", e)
                continue
            
            # Interpolate model output to experimental x points (convert x_exp_mm to meters)
            x_exp_profile_m = x_exp_profile_mm * 1e-3
            w_interp = tf_linear_interp(x_model_m, w_model, x_exp_profile_m)

            # Calculate normalized loss
            w_exp_norm = (w_exp_profile - w_exp_mean) / (w_exp_std + 1e-12)
            w_interp_norm = (w_interp - w_exp_mean) / (w_exp_std + 1e-12)

            current_profile_loss = tf.reduce_mean(tf.square(w_interp_norm - w_exp_norm))

            if tf.math.is_nan(current_profile_loss):
                # tf.print("⚠️ NaN in loss for current profile, skipping.")
                continue

            total_batch_loss += current_profile_loss
            valid_profile_count += 1
            
    # Calculate mean loss for the batch
    if valid_profile_count > 0:
        mean_batch_loss = total_batch_loss / tf.cast(valid_profile_count, tf.float32)
    else:
        # If no valid profiles in batch, return a large loss to signal failure
        mean_batch_loss = tf.constant(1e5, dtype=tf.float32) 
    
    # Compute gradients for all variables (models + calibrators)
    grads = tape.gradient(mean_batch_loss, all_trainable_variables)
    
    # Clip and clean gradients
    grads_clean = []
    for g in grads:
        if g is not None:
            g = tf.clip_by_value(g, -0.5, 0.5) # Clip to prevent exploding gradients
            if tf.reduce_any(tf.math.is_nan(g)):
                # tf.print("⚠️ NaN in gradients, replacing with zeros.")
                grads_clean.append(tf.zeros_like(g))
            else:
                grads_clean.append(g)
        else:
            grads_clean.append(None) # Keep None for non-existent gradients

    # Only apply gradients if at least one is not None
    if any(g is not None for g in grads_clean):
        optimizer.apply_gradients(zip(grads_clean, all_trainable_variables))
    # else: skip optimizer step for this batch

    return mean_batch_loss, valid_profile_count

# === 9. Visualization function for time-dependent parameters (NO CHANGE TO LOGIC) ===
def visualize_time_dependent_params(input_features, title_suffix=""):
    time_points = np.linspace(0, 1, 100)
    D_values = []
    alpha_values = []
    n_values = []
    
    # Get base scales for this specific input_features once
    input_features_normalized = normalize_input_features(tf.reshape(input_features, (1, -1)))
    base_scales = main_model(input_features_normalized)[0]

    for t in time_points:
        t_norm = tf.constant(t, dtype=tf.float32)
        Dv, alpha0, nShape = get_time_dependent_params(t_norm, base_scales)
        D_values.append(Dv.numpy())
        alpha_values.append(alpha0.numpy())
        n_values.append(nShape.numpy())
    
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    
    axes[0].plot(time_points, D_values, 'b-', linewidth=2)
    axes[0].set_xlabel('Normalized Time')
    axes[0].set_ylabel('D (m²/s)')
    axes[0].set_title(f'D(t) {title_suffix}')
    axes[0].grid(True)
    axes[0].ticklabel_format(style='sci', axis='y', scilimits=(0,0))
    
    axes[1].plot(time_points, alpha_values, 'g-', linewidth=2)
    axes[1].set_xlabel('Normalized Time')
    axes[1].set_ylabel('α')
    axes[1].set_title(f'α(t) {title_suffix}')
    axes[1].grid(True)
    
    axes[2].plot(time_points, n_values, 'r-', linewidth=2)
    axes[2].set_xlabel('Normalized Time')
    axes[2].set_ylabel('n')
    axes[2].set_title(f'n(t) {title_suffix}')
    axes[2].grid(True)
    
    plt.tight_layout()
    plt.show()

# === 10. Training loop ===
loss_history = []
param_history = {'Dv_mean': [], 'alpha_mean': [], 'n_mean': []} # To store mean params from each epoch

print("🚀 Starting training with time-dependent parameters (with calibrators)...")

for epoch in range(EPOCHS):
    epoch_start_time = time.time()
    
    total_epoch_loss = 0.0
    total_valid_profiles_in_epoch = 0
    param_epoch_vals = {'Dv_mean': [], 'alpha_mean': [], 'n_mean': []} # Reset for current epoch

    for batch_inputs, batch_x_exp_mm, batch_w_exp in full_dataset:
        batch_loss, valid_count_in_batch = train_step(batch_inputs, batch_x_exp_mm, batch_w_exp)
        
        total_epoch_loss += batch_loss * tf.cast(valid_count_in_batch, tf.float32) # Scale by valid count
        total_valid_profiles_in_epoch += valid_count_in_batch

        # Collect params for mean calculation in this epoch (from the first profile in batch)
        # This is a heuristic for visualization, true mean would be over all profiles.
        if valid_count_in_batch > 0:
            sample_input_features = batch_inputs[0] # Take first valid profile's input features
            normalized_sample_input = normalize_input_features(tf.reshape(sample_input_features, (1, -1)))
            sample_base_scales = main_model(normalized_sample_input)[0]

            for t_sample in [0.0, 0.5, 1.0]: # Sample 3 points in time
                t_norm = tf.constant(t_sample)
                Dv_t, alpha_t, n_t = get_time_dependent_params(t_norm, sample_base_scales)
                param_epoch_vals['Dv_mean'].append(Dv_t.numpy())
                param_epoch_vals['alpha_mean'].append(alpha_t.numpy())
                param_epoch_vals['n_mean'].append(n_t.numpy())

    if total_valid_profiles_in_epoch == 0:
        print(f"Epoch {epoch}: ❌ No valid profiles processed → training stopped.")
        break
    
    mean_epoch_loss = total_epoch_loss / tf.cast(total_valid_profiles_in_epoch, tf.float32)
    loss_history.append(mean_epoch_loss.numpy())

    mean_params = {k: np.mean(v) if v else 0.0 for k, v in param_epoch_vals.items()}
    for key, value in mean_params.items():
        param_history[key].append(value)

    epoch_end_time = time.time()

    if epoch % 1 == 0: # Print every epoch
        print(f"\n📉 Epoch {epoch} took {epoch_end_time - epoch_start_time:.2f}s, Loss: {mean_epoch_loss:.4e}")
        print(f"   Processed {total_valid_profiles_in_epoch} profiles. Mean values: D={mean_params['Dv_mean']:.2e}, α={mean_params['alpha_mean']:.4f}, n={mean_params['n_mean']:.2f}")
        
        # Also print calibrator values to see if they are changing
        # Ensure they are not None before accessing .numpy()
        cal_D_scale = _D_calibrator_scale.numpy() if _D_calibrator_scale is not None else 'N/A'
        cal_D_offset = _D_calibrator_offset.numpy() if _D_calibrator_offset is not None else 'N/A'
        cal_alpha_scale = _alpha_calibrator_scale.numpy() if _alpha_calibrator_scale is not None else 'N/A'
        cal_alpha_offset = _alpha_calibrator_offset.numpy() if _alpha_calibrator_offset is not None else 'N/A'
        cal_n_scale = _n_calibrator_scale.numpy() if _n_calibrator_scale is not None else 'N/A'
        cal_n_offset = _n_calibrator_offset.numpy() if _n_calibrator_offset is not None else 'N/A'

        print(f"   Calibrators: D_scale={cal_D_scale:.4f}, D_offset={cal_D_offset:.4f}, "
              f"alpha_scale={cal_alpha_scale:.4f}, alpha_offset={cal_alpha_offset:.4f}, "
              f"n_scale={cal_n_scale:.4f}, n_offset={cal_n_offset:.4f}")

        if epoch % 10 == 0: # Visualize less frequently
            # Pick a representative input for visualization
            test_input_for_plot = tf.constant([0.119, 20.0, 0.45, 0.6, 72.0], dtype=tf.float32)
            visualize_time_dependent_params(test_input_for_plot, 
                                            title_suffix=f"(Epoch {epoch})")
            
            # Save weights (optional)
            # main_model.save_weights(f"main_model_timedep_epoch{epoch}.weights.h5")
            # D_network.save_weights(f"D_network_epoch{epoch}.weights.h5")
            # alpha_network.save_weights(f"alpha_network_epoch{epoch}.weights.h5")
            # n_network.save_weights(f"n_network_epoch{epoch}.weights.h5")

# === 11. Final convergence plots ===
if loss_history:
    fig, axes = plt.subplots(2, 2, figsize=(12, 8))

    axes[0,0].plot(loss_history, label="Loss", linewidth=2, color='red')
    axes[0,0].set_xlabel("Epoch")
    axes[0,0].set_ylabel("Loss")
    axes[0,0].set_title("Loss Convergence")
    axes[0,0].grid(True)
    axes[0,0].legend()
    axes[0,0].set_yscale('log')

    # Ensure param_history is not empty before plotting
    if param_history['Dv_mean'] and any(v != 0.0 for v in param_history['Dv_mean']):
        axes[0,1].plot(param_history['Dv_mean'], label="D mean", linewidth=2, color='blue')
        axes[0,1].set_xlabel("Epoch")
        axes[0,1].set_ylabel("D (m²/s)")
        axes[0,1].set_title("Mean D Evolution")
        axes[0,1].grid(True)
        axes[0,1].legend()

    if param_history['alpha_mean'] and any(v != 0.0 for v in param_history['alpha_mean']):
        axes[1,0].plot(param_history['alpha_mean'], label="α mean", linewidth=2, color='green')
        axes[1,0].set_xlabel("Epoch")
        axes[1,0].set_ylabel("α")
        axes[1,0].set_title("Mean α Evolution")
        axes[1,0].grid(True)
        axes[1,0].legend()

    if param_history['n_mean'] and any(v != 0.0 for v in param_history['n_mean']):
        axes[1,1].plot(param_history['n_mean'], label="n mean", linewidth=2, color='orange')
        axes[1,1].set_xlabel("Epoch")
        axes[1,1].set_ylabel("n")
        axes[1,1].set_title("Mean n Evolution")
        axes[1,1].grid(True)
        axes[1,1].legend()

    plt.tight_layout()
    plt.show()

print("🎉 Training completed! Model with calibrators.")

# === 12. Test the trained model ===
print("\n🧪 Testing time-dependent model (with calibrators)...")
# Using a representative input for testing
test_input_features = tf.constant([0.119, 20.0, 0.45, 0.6, 72.0], dtype=tf.float32)

visualize_time_dependent_params(test_input_features, 
                                title_suffix="(Final Model with Calibrators)")

# Simulate with the final trained model
x_model_m_test, w_model_test = simulate_model_tf_time_dependent(
    test_input_features, D_network, alpha_network, n_network,
    dx=1e-3, dt=30.0, Tmax=72*3600, Nx=100, picard_iterations=PICARD_ITERATIONS # Use more detailed Nx for final plot
)

plt.figure(figsize=(8, 5))
plt.plot(x_model_m_test.numpy()*1000, w_model_test.numpy(), label="Simulation with time-dependent params + calibrators", linewidth=2)
plt.title("Test Case - t=72h with Calibrated Time-Dependent Parameters")
plt.xlabel("x [mm]")
plt.ylabel("w [kg/m³]")
plt.grid()
plt.legend()
plt.show()
```
